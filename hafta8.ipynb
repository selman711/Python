```python
width, height, channel=300,300,3
img_red = np.zeros((width,height,channel),dtype=np.uint8)
img_red[:,:,0]=255
plt.imshow(img_red)
plt.axis("off")
```

```python
width, height, channel=300,300,3
img_green = np.zeros((width,height,channel),dtype=np.uint8)
img_green[:,:,1]=255
plt.imshow(img_green)
plt.axis("off")
```

```python
width, height, channel=300,300,3
img_blue = np.zeros((width,height,channel),dtype=np.uint8)
img_blue[:,:,2]=255
plt.imshow(img_blue)
plt.axis("off")
```

```python
fig=plt.figure(figsize=(10,7))
rows=1
colums=3

fig.add_subplot(rows, colums,1)
plt.imshow(img_red)
plt.axis("off")
plt.title("RED Image")

fig.add_subplot(rows, colums,2)
plt.imshow(img_green)
plt.axis("off")
plt.title("GREEN Image")

fig.add_subplot(rows, colums,3)
plt.imshow(img_blue)
plt.axis("off")
plt.title("BLUE Image")
```

```python
img_black=np.zeros((width,height,3),dtype=np.uint8)
img_white=np.ones((width,height,3),dtype=np.uint8)*255

img_gray_1 = np.zeros((width,height,3),dtype=np.uint8)
img_gray_1[:,:,:] =128

img_gray_2= np.random.randint(128,129,(width,height,3),dtype=np.uint8)
```

```python
fig=plt.figure(figsize=(10,7),facecolor='green')
rows=2
colums=2

fig.add_subplot(rows, colums,1)
plt.imshow(img_white)
plt.axis("off")
plt.title("WHİTE Image")

fig.add_subplot(rows, colums,2)
plt.imshow(img_black)
plt.axis("off")
plt.title("BLACK Image")

fig.add_subplot(rows, colums,3)
plt.imshow(img_gray_1)
plt.axis("off")
plt.title("GRAY-1 Image")

fig.add_subplot(rows, colums,4)
plt.imshow(img_gray_2)
plt.axis("off")
plt.title("GRAY-2 Image")

```

```python
width,height,channel =300,420,3
img_flag_germany=np.zeros((width,height,channel),dtype=np.uint8)
img_flag_germany[:100,:]=[0,0,0]
img_flag_germany[100:200,:]=[255,0,0]
img_flag_germany[200:300,:]=[255,255,0]
plt.imshow(img_flag_germany)
plt.axis("off")
plt.show()
```

```python
width,height,channel =300,420,3
img_flag_russian=np.zeros((width,height,channel),dtype=np.uint8)
img_flag_russian[:100,:]=[255,255,255]
img_flag_russian[100:200,:]=[0,0,255]
img_flag_russian[200:300,:]=[255,0,0]
plt.imshow(img_flag_russian)
plt.axis("off")
plt.show()
```

```python
width,height,channel =300,450,3
img_flag_belgium=np.zeros((width,height,channel),dtype=np.uint8)
img_flag_belgium[:,:150]=[0,0,0]
img_flag_belgium[:,150:300]=[255,255,0]
img_flag_belgium[:,300:450]=[255,0,0]
plt.imshow(img_flag_belgium)
plt.axis("off")
plt.show()
```

```python
mask_red=(img_flag_russian[:,:,0]==255) & (img_flag_russian[:,:,1]==0) &(img_flag_russian[:,:,1]==0)
img_flag_russian[mask_red] =[255,255,0]
plt.imshow(img_flag_russian)
plt.axis("off")
```
